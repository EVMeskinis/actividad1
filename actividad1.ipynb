# Actividad 1 – Organizando información con estructuras de datos
El código está en `src/`. Este notebook solo importa las funciones y las prueba.

```
from src.datos import COLUMNAS, ROLES
from src.informe import generar_informe, mostrar_informe
```

## Sin rol: todas las columnas por completitud descendente

```
mostrar_informe(generar_informe(), "Todas las columnas")
```

## Informe por cada rol

```
for rol in ROLES:
    mostrar_informe(generar_informe(rol), f"Rol: {rol}")
```

## Casos de prueba de errores

```
# Rol que no existe
mostrar_informe(generar_informe("gerente"), "Rol inexistente")
```

```
# Rol con criterio inválido y una columna que no existe (sin tocar datos.py)
roles_prueba = {
    "prueba": {"columnas": ["EDAD", "ITF", "SEXO"], "criterio": "promedio", "orden": "A"}
}
mostrar_informe(generar_informe("prueba", roles=roles_prueba), "Rol de prueba")
```
